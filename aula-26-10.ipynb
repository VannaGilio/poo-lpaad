```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

    def mostrar_nome (self):
        print(f"Nome da músga {self.nome}")
    def mostrar_cantor (self):
        print(f"Nome do cantor {self.nome}")
    def mostrar_genero (self):
        print(f"Genero da músga {self.nome}")
    def mostrar_duracao (self):
        print(f"Duração da músga {self.nome}")
    def mostrar_ano (self):
        print(f"Ano de lançamento da músga {self.nome}")
    def mostrar_faixa (self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()

    #################

    def reproduzir(self):
        self.reproduzir += 1   
        print(f'Reproduzido: {self.nome}')

m1 = Musica("Magica", "Calcinha Preta", "Rock", 43, 2001)

m1.mostrar_faixa()
```
